```python
"""Random seed 고정 유틸리티.

실험을 재현하려면 제일 먼저 시드를 고정해야 한다.
다음 네 곳을 모두 같은 값으로 고정해야 실제로 동일한 결과가 나온다:
  1. Python 기본 random 모듈
  2. NumPy
  3. PyTorch (CPU + GPU)
  4. PYTHONHASHSEED (환경 변수)

cuDNN deterministic 모드를 켜면 속도는 약간 느리지만 원인 모든 랜덜성이 제거된다.
"""
import os
import random
import numpy as np
import torch
```

## Step 2. Function: set_seed

이 셀은 원본 코드의 해당 블록을 그대로 옮긴 단계입니다.


```python
def set_seed(seed: int = 42) -> None:
    """seed 값을 받아 여러 라이브러리의 랜덜 상태를 일일이 고정한다.

    사용 예시::

        set_seed(42)  # 모든 스크립트 맨 위에서 호출
    """
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    os.environ["PYTHONHASHSEED"] = str(seed)
    # benchmark=False : 매 실행마다 최적 커널을 지정하지 않아 결과가 일정해짐
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False
```
