```python
from pathlib import Path
import os
import sys

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / 'src').exists():
    # 노트북이 다른 경로에서 열렸을 때 프로젝트 루트 자동 탐색
    for parent in [Path.cwd(), *Path.cwd().parents]:
        if (parent / 'src').exists() and (parent / 'configs').exists():
            PROJECT_ROOT = parent
            break
os.chdir(PROJECT_ROOT)
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))
print(f'Project root: {PROJECT_ROOT}')
```

## Step 1. Setup and Imports

이 셀은 원본 코드의 해당 블록을 그대로 옮긴 단계입니다.


```python
"""YAML config 로더.

cofig.yaml 파일을 Python dict로 변환해주는 파일.
load_config('configs/default.yaml') 한 줄로 학습/평가/추론 스크립트들이
설정값을 불러온다.
"""
from pathlib import Path
import yaml
```

## Step 2. Function: load_config

이 셀은 원본 코드의 해당 블록을 그대로 옮긴 단계입니다.


```python
def load_config(path: str | Path) -> dict:
    """.yaml 파일을 읽어 Python dict로 반환한다.

    사용 예시::

        cfg = load_config('configs/default.yaml')
        print(cfg['train']['lr'])   # 0.001
        print(cfg['model']['name']) # efficientnet_b0
    """
    with open(path, "r", encoding="utf-8") as f:
        return yaml.safe_load(f)
```
