## Step 1. Setup and Imports

이 셀은 원본 코드의 해당 블록을 그대로 옮긴 단계입니다.


```python
"""TensorBoard 로거 래퍼.

학습 중 loss, accuracy, F1 등의 수치 변화를 TensorBoard에 기록한다.
학습이 끝난 후 runs/ 폴더에 로그가 생성되며,
아래 명령어로 브라우저에서 확인할 수 있다:

    tensorboard --logdir runs
"""
from pathlib import Path
from datetime import datetime
from torch.utils.tensorboard import SummaryWriter
```

## Step 2. Class: TBLogger

이 셀은 원본 코드의 해당 블록을 그대로 옮긴 단계입니다.


```python
class TBLogger:
    """SummaryWriter를 람핑한 간단 로거.

    사용 예시::

        logger = TBLogger('runs')
        logger.log({'train/loss': 0.4, 'val/f1': 0.87}, step=1)
        logger.close()
    """
```

## Step 3. Function: __init__

이 셀은 원본 코드의 해당 블록을 그대로 옮긴 단계입니다.


```python
    def __init__(self, log_dir: str, run_name: str | None = None):
        # run_name을 지정하지 않으면 현재 시간으로 자동 생성 (runs/20260423_120000)
        run_name = run_name or datetime.now().strftime("%Y%m%d_%H%M%S")
        self.path = Path(log_dir) / run_name
        self.path.mkdir(parents=True, exist_ok=True)
        self.writer = SummaryWriter(self.path.as_posix())
```

## Step 4. Function: log

이 셀은 원본 코드의 해당 블록을 그대로 옮긴 단계입니다.


```python
    def log(self, scalars: dict, step: int) -> None:
        # scalars = {'train/loss': 0.4, 'val/acc': 0.93} 형태로 넘기면
        # TensorBoard에 한 번에 여러 항목을 기록한다.
        for k, v in scalars.items():
            self.writer.add_scalar(k, v, step)
```

## Step 5. Function: close

이 셀은 원본 코드의 해당 블록을 그대로 옮긴 단계입니다.


```python
    def close(self) -> None:
        # 학습 후 반드시 close()을 호출해야 로그가 정상적으로 닫히는 것이 보장된다.
        self.writer.close()
```
