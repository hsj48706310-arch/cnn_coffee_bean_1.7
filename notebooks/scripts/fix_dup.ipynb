# Fix Dup Notebook

- Source: `scripts/fix_dup.py`
- 목적: 원본 파이썬 파일을 단계별로 실행/설명하기 위한 노트북 버전
- 실행 방법: 위에서 아래로 순서대로 실행


```python
from pathlib import Path
import os
import sys

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / 'src').exists():
    # 노트북이 다른 경로에서 열렸을 때 프로젝트 루트 자동 탐색
    for parent in [Path.cwd(), *Path.cwd().parents]:
        if (parent / 'src').exists() and (parent / 'configs').exists():
            PROJECT_ROOT = parent
            break
os.chdir(PROJECT_ROOT)
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))
print(f'Project root: {PROJECT_ROOT}')
```

## Step 1. Full Script

이 셀은 원본 코드의 해당 블록을 그대로 옮긴 단계입니다.


```python
import sys
p = 'app/streamlit_app.py'
t = open(p, encoding='utf-8').read()
positions = []
i = 0
while True:
    j = t.find('from __future__', i)
    if j < 0:
        break
    positions.append(j)
    i = j + 1
print('from __future__ positions:', positions)
print('total len:', len(t))
if len(positions) >= 2:
    new = t[:positions[1]]
    open(p, 'w', encoding='utf-8', newline='\n').write(new)
    print('rewritten, new len:', len(new))
```
